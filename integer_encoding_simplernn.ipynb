```python
import numpy as np
```

```python
docs = ['go india',
		'india india',
		'hip hip hurray',
		'jeetega bhai jeetega india jeetega',
		'bharat mata ki jai',
		'kohli kohli',
		'sachin sachin',
		'dhoni dhoni',
		'modi ji ki jai',
		'inquilab zindabad']
```

```python
!pip install tensorflow
from tensorflow.keras.preprocessing.text import Tokenizer
tokenizer = Tokenizer(oov_token = 'kowshiq')
```

```python
tokenizer.fit_on_texts(docs)
```

```python
sequences = tokenizer.texts_to_sequences(docs)
sequences
```

```python
from keras.utils import pad_sequences
```

```python
sequences = pad_sequences(sequences, padding = 'post')
```

```python
sequences
```

```python
from keras.datasets import imdb
from keras import Sequential
from keras.layers import Dense, SimpleRNN, Embedding, Flatten
```

```python
(X_train, y_train), (X_test, y_test) = imdb.load_data()
```

```python
X_train  # already preprocessed
```

```python
X_train = pad_sequences(X_train, padding = 'post', maxlen = 50)
X_test = pad_sequences(X_test, padding = 'post', maxlen = 50)
```

```python
X_train.shape
```

```python
model = Sequential()

model.add(SimpleRNN(32, input_shape = (50, 1), return_sequences = False)) #the output at each layer will not come out if we put return sequence as false
model.add(Dense(1, activation = 'sigmoid'))

model.summary()
```

image.png

```python
model.compile(loss = 'binary_crossentropy', optimizer = 'adam', metrics = ['accuracy'])

model.fit(X_train, y_train, epochs = 5, validation_data = (X_test, y_test))
```
